```python
# 导入必要的库
import pandas as pd
import os

# 显示当前工作目录
print(f"当前工作目录: {os.getcwd()}")

# 假设CSV文件在项目根目录或者特定目录下
# 修改路径以指向你的实际CSV文件
file_path = "/media/marxin/softs/workspace/Whisper_of_DNA_pl/data/example-data/maize1404/SNP_pca/1404cubic.csv"  # 根据实际文件位置调整路径

try:
    # 使用pandas读取CSV文件
    df = pd.read_csv(file_path)
    
    # 显示CSV文件的基本信息
    print(f"CSV文件形状: {df.shape}")
    print("\n前5行数据:")
    print(df.head())
    
    # 显示列名
    print("\n列名:")
    print(df.columns.tolist())
    
    # 检查数据类型
    print("\n数据类型:")
    print(df.dtypes)
    
    # 检查是否有缺失值
    print("\n缺失值统计:")
    print(df.isnull().sum())
    
except FileNotFoundError:
    print(f"文件未找到: {file_path}")
    print("请确认文件路径是否正确，或者提供以下可能的CSV文件列表:")
    
    # 列出当前目录和上级目录中的CSV文件
    csv_files = []
    
    # 检查当前目录
    for root, dirs, files in os.walk('.'):
        for file in files:
            if file.endswith('.csv'):
                csv_files.append(os.path.join(root, file))
    
    # 检查上级目录
    for root, dirs, files in os.walk('..'):
        for file in files:
            if file.endswith('.csv'):
                csv_files.append(os.path.join(root, file))
    
    if csv_files:
        print("找到的CSV文件:")
        for file in csv_files:
            print(f"- {file}")
    else:
        print("未在当前目录或上级目录找到CSV文件")
except Exception as e:
    print(f"读取CSV文件时出错: {e}")
```
