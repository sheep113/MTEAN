```python
# 导入必要的库
import os
import pandas as pd
from pathlib import Path

def merge_files(folder_path, output_path=None, delimiter=","):
    """
    将指定文件夹中的所有文本文件按列合并成一个CSV文件
    
    Args:
        folder_path: 包含要合并文件的文件夹路径
        output_path: 输出文件路径，默认为folder_path/merged.csv
        delimiter: 输出CSV文件的分隔符
    """
    if output_path is None:
        output_path = os.path.join(folder_path, "merged.csv")
    
    # 获取文件夹中的所有文本文件
    files = [f for f in os.listdir(folder_path) if os.path.isfile(os.path.join(folder_path, f))]
    
    if not files:
        print(f"警告: 文件夹 {folder_path} 中没有找到文件")
        return
    
    # 用于存储所有文件的数据
    data_dict = {}
    max_length = 0
    
    # 读取每个文件并提取数据
    for file_name in files:
        file_path = os.path.join(folder_path, file_name)
        try:
            with open(file_path, 'r') as f:
                lines = f.read().strip().split('\n')
                
                # 第一行作为列名
                column_name = lines[0]
                
                # 剩余行作为数据
                values = lines[1:]
                
                # 更新最大长度
                max_length = max(max_length, len(values))
                
                # 将数据添加到字典中
                data_dict[column_name] = values
                
                print(f"已读取 {file_name}, 列名: {column_name}, 行数: {len(values)}")
        except Exception as e:
            print(f"无法读取文件 {file_name}: {e}")
    
    # 确保所有列的长度相同
    for column_name, values in data_dict.items():
        if len(values) < max_length:
            # 对于短列，用空值填充
            data_dict[column_name] = values + [''] * (max_length - len(values))
    
    # 创建DataFrame
    df = pd.DataFrame(data_dict)
    
    # 保存到CSV文件
    df.to_csv(output_path, index=False, sep=delimiter)
    print(f"已合并 {len(data_dict)} 个文件到 {output_path}")
    print(f"合并后的文件包含 {max_length} 行数据")
    
    # 返回DataFrame以便在Notebook中进一步分析
    return df
```

```python
# 调用示例：合并SNP_pca文件夹中的所有文件
example_folder = "/media/marxin/softs/workspace/Whisper_of_DNA_pl/data/example-data/maize1404/maize1404-F1/SNP_pca"
output_file = "/media/marxin/softs/workspace/Whisper_of_DNA_pl/data/example-data/maize1404/maize1404-F1/merged_snp_data.csv"

# 调用合并函数
df = merge_files(example_folder, output_file)

# 显示合并后的数据前5行
df.head()
```

```python
# 导入必要的库
import os
import pandas as pd
from pathlib import Path

def merge_files_enhanced(folder_path, output_path=None, delimiter=",", file_extensions=None, ignore_files=None):
    """
    将指定文件夹中的所有文本文件按列合并成一个CSV文件（增强版）
    
    Args:
        folder_path: 包含要合并文件的文件夹路径
        output_path: 输出文件路径，默认为folder_path/merged.csv
        delimiter: 输出CSV文件的分隔符
        file_extensions: 要处理的文件扩展名列表，例如['.txt', '.csv']，默认为None（处理所有文件）
        ignore_files: 要忽略的文件名列表，默认为None
    """
    if output_path is None:
        output_path = os.path.join(folder_path, "merged.csv")
    
    # 获取文件夹中的所有文件
    all_files = [f for f in os.listdir(folder_path) if os.path.isfile(os.path.join(folder_path, f))]
    
    # 根据扩展名和忽略列表筛选文件
    files = []
    for f in all_files:
        if ignore_files and f in ignore_files:
            continue
        if file_extensions:
            if any(f.endswith(ext) for ext in file_extensions):
                files.append(f)
        else:
            files.append(f)
    
    if not files:
        print(f"警告: 文件夹 {folder_path} 中没有找到符合条件的文件")
        return
    
    # 用于存储所有文件的数据
    data_dict = {}
    max_length = 0
    
    # 读取每个文件并提取数据
    for file_name in sorted(files):  # 排序确保文件处理顺序一致
        file_path = os.path.join(folder_path, file_name)
        try:
            with open(file_path, 'r') as f:
                lines = f.read().strip().split('\n')
                
                if not lines:
                    print(f"警告: 文件 {file_name} 为空")
                    continue
                
                # 第一行作为列名
                column_name = lines[0]
                
                # 如果列名已存在，用文件名作为前缀
                if column_name in data_dict:
                    column_name = f"{os.path.splitext(file_name)[0]}_{column_name}"
                
                # 剩余行作为数据
                values = lines[1:]
                
                # 更新最大长度
                max_length = max(max_length, len(values))
                
                # 将数据添加到字典中
                data_dict[column_name] = values
                
                print(f"已读取 {file_name}, 列名: {column_name}, 行数: {len(values)}")
        except Exception as e:
            print(f"无法读取文件 {file_name}: {e}")
    
    # 确保所有列的长度相同
    for column_name, values in data_dict.items():
        if len(values) < max_length:
            # 对于短列，用空值填充
            data_dict[column_name] = values + [''] * (max_length - len(values))
    
    # 创建DataFrame
    df = pd.DataFrame(data_dict)
    
    # 保存到CSV文件
    df.to_csv(output_path, index=False, sep=delimiter)
    print(f"已合并 {len(data_dict)} 个文件到 {output_path}")
    print(f"合并后的文件包含 {max_length} 行数据")
    
    return df
```

```python
# 调用增强版示例
example_folder = "/media/marxin/softs/workspace/Whisper_of_DNA_pl/data/example-data/wheat599/SNP_origin/"
output_file = "/media/marxin/softs/workspace/Whisper_of_DNA_pl/data/example-data/wheat599/SNP_origin/wheatY.csv"

# 只处理.txt文件，忽略特定文件
df_enhanced = merge_files_enhanced(
    example_folder, 
    output_file, 
    file_extensions=['.Y'], 
    ignore_files=['cubicf1_pc95.csv', 'cubicf1_pc95.pkl']
)

# 显示合并后的数据
df_enhanced.head()
```
