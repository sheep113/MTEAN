```python
import pandas as pd
import matplotlib.pyplot as plt
import os
import numpy as np
from scipy.stats import pearsonr
import scipy.stats as stats
import seaborn as sns
from scipy import stats as scipy_stats

# Set publication-ready style
plt.style.use('seaborn-v0_8-whitegrid')
sns.set_palette("husl")

# Configure matplotlib for publication quality
plt.rcParams.update({
    'font.size': 12,
    'font.family': 'serif',
    'font.serif': ['Times New Roman'],
    'axes.labelsize': 14,
    'axes.titlesize': 16,
    'xtick.labelsize': 12,
    'ytick.labelsize': 12,
    'legend.fontsize': 12,
    'figure.titlesize': 18,
    'lines.linewidth': 2,
    'axes.linewidth': 1.2,
    'grid.alpha': 0.3,
    'axes.spines.top': False,
    'axes.spines.right': False
})

# Define file paths for pre-trained model only
log_dirfinetunes = '../logs0/DNAWhisperfinetune/run0_wheat_2000_phonall__30k/'
# log_dirfinetunes_wheat = '../logs0/DNAWhisperfinetune/run0_wheat_599_envall_finetune_30k/'

# Load datasets for pre-trained model
pretrained_datasets = {}
colors = {'val': '#2E86AB', 'test': '#A23B72'}  # Professional color scheme
split_names = {'val': 'Validation', 'test': 'Test'}

try:
    val_file = os.path.join(log_dirfinetunes, 'val_predictions.csv')
    test_file = os.path.join(log_dirfinetunes, 'test_predictions.csv')
    
    pretrained_datasets = {
        'val': pd.read_csv(val_file),
        'test': pd.read_csv(test_file)
    }
    print("Successfully loaded pre-trained model datasets")
    print(f"Val dataset shape: {pretrained_datasets['val'].shape}")
    print(f"Test dataset shape: {pretrained_datasets['test'].shape}")
except FileNotFoundError as e:
    print(f"Error loading pre-trained model files: {e}")
    pretrained_datasets = {'val': pd.DataFrame(), 'test': pd.DataFrame()}

# Get phenotypes from the datasets
phenotypes = []
for split_name, df in pretrained_datasets.items():
    if not df.empty:
        phenotypes = sorted(list(set([col.split('_')[0] for col in df.columns if '_' in col])))
        break

print(f"Found phenotypes: {phenotypes}")

def compare_val_test_analysis_publication(datasets, phenotypes):
    """
    Publication-ready comparison analysis of validation and test sets for pre-trained model
    Each plot is generated separately for easy integration into papers
    """
    
    for pheno in phenotypes:
        pred_col = f"{pheno}_pred"
        label_col = f"{pheno}_label"
        
        # Check if both splits have data for this phenotype
        residuals_data = {}
        for split in ['val', 'test']:
            df = datasets[split]
            if not df.empty and pred_col in df.columns and label_col in df.columns:
                residuals = df[pred_col] - df[label_col]
                if not residuals.empty and residuals.nunique() >= 2:
                    residuals_data[split] = residuals
        
        if len(residuals_data) < 2:
            print(f"Skipping phenotype {pheno}: Insufficient data")
            continue
            
        print(f"\n=== Phenotype {pheno.upper()} Comparison Analysis (Validation vs Test) ===")
        
        # 1. Q-Q Plot Comparison
        fig1, ax1 = plt.subplots(figsize=(8, 6))
        
        # Get theoretical quantiles for the smaller dataset
        sample_size = min(len(residuals_data['val']), len(residuals_data['test']))
        theoretical_quantiles = scipy_stats.norm.ppf(np.linspace(0.01, 0.99, sample_size))
        
        # Plot sample quantiles for both splits
        for split, residuals in residuals_data.items():
            sample_quantiles = np.sort(residuals)
            # Interpolate to match theoretical quantiles length
            if len(sample_quantiles) != len(theoretical_quantiles):
                x_interp = np.linspace(0, 1, len(sample_quantiles))
                x_new = np.linspace(0, 1, len(theoretical_quantiles))
                sample_quantiles = np.interp(x_new, x_interp, sample_quantiles)
            
            ax1.scatter(theoretical_quantiles, sample_quantiles, 
                       color=colors[split], alpha=0.7, s=25,
                       label=split_names[split], edgecolors='white', linewidth=0.5)
        
        # Add reference line for normal distribution
        min_val = min(theoretical_quantiles.min(), 
                     min([np.sort(res)[0] for res in residuals_data.values()]))
        max_val = max(theoretical_quantiles.max(), 
                     max([np.sort(res)[-1] for res in residuals_data.values()]))
        ax1.plot([min_val, max_val], [min_val, max_val], 'k--', 
                alpha=0.8, linewidth=2, label='Normal Distribution')
        
        ax1.set_title(f'Q-Q Plot for {pheno.upper()} (Validation vs Test)', fontweight='bold', pad=20)
        ax1.set_xlabel('Theoretical Quantiles', fontweight='bold')
        ax1.set_ylabel('Sample Quantiles', fontweight='bold')
        ax1.legend(frameon=True, fancybox=True, shadow=True)
        ax1.grid(True, alpha=0.3)
        plt.tight_layout()
        plt.show()
        
        # 2. Histogram Comparison with KDE
        fig2, ax2 = plt.subplots(figsize=(8, 6))
        
        for split, residuals in residuals_data.items():
            ax2.hist(residuals, bins=30, alpha=0.6, color=colors[split], 
                    label=split_names[split], density=True, edgecolor='white', linewidth=0.7)
            
            # Add KDE overlay
            kde_x = np.linspace(residuals.min(), residuals.max(), 100)
            kde = scipy_stats.gaussian_kde(residuals)
            ax2.plot(kde_x, kde(kde_x), color=colors[split], linewidth=3, alpha=0.9)
        
        ax2.set_title(f'Residual Distribution for {pheno.upper()} (Validation vs Test)', 
                     fontweight='bold', pad=20)
        ax2.set_xlabel('Residual Value', fontweight='bold')
        ax2.set_ylabel('Density', fontweight='bold')
        ax2.legend(frameon=True, fancybox=True, shadow=True)
        ax2.grid(True, alpha=0.3)
        plt.tight_layout()
        plt.show()
        
        # 3. Box Plot Comparison
        fig3, ax3 = plt.subplots(figsize=(8, 6))
        
        box_data = [residuals_data['val'], residuals_data['test']]
        box_labels = ['Validation', 'Test']
        
        box_plot = ax3.boxplot(box_data, labels=box_labels, patch_artist=True,
                              boxprops=dict(linewidth=1.5),
                              medianprops=dict(linewidth=2, color='black'),
                              whiskerprops=dict(linewidth=1.5),
                              capprops=dict(linewidth=1.5))
        
        # Set box colors
        box_plot['boxes'][0].set_facecolor(colors['val'])
        box_plot['boxes'][0].set_alpha(0.7)
        box_plot['boxes'][1].set_facecolor(colors['test'])
        box_plot['boxes'][1].set_alpha(0.7)
        
        ax3.set_title(f'Box Plot Comparison for {pheno.upper()} (Validation vs Test)', 
                     fontweight='bold', pad=20)
        ax3.set_ylabel('Residual Value', fontweight='bold')
        ax3.grid(True, alpha=0.3, axis='y')
        plt.tight_layout()
        plt.show()
        
        # 4. Kernel Density Estimation Comparison
        fig4, ax4 = plt.subplots(figsize=(8, 6))
        
        for split, residuals in residuals_data.items():
            sns.kdeplot(data=residuals, ax=ax4, color=colors[split], 
                       label=split_names[split], linewidth=3, fill=True, alpha=0.3)
        
        ax4.set_title(f'Kernel Density Estimation for {pheno.upper()} (Validation vs Test)', 
                     fontweight='bold', pad=20)
        ax4.set_xlabel('Residual Value', fontweight='bold')
        ax4.set_ylabel('Density', fontweight='bold')
        ax4.legend(frameon=True, fancybox=True, shadow=True)
        ax4.grid(True, alpha=0.3)
        plt.tight_layout()
        plt.show()
        
        # 5. Violin Plot Comparison
        fig5, ax5 = plt.subplots(figsize=(8, 6))
        
        # Prepare data for violin plot
        violin_data = []
        violin_labels = []
        for split, residuals in residuals_data.items():
            violin_data.append(residuals)
            violin_labels.append(split_names[split])
        
        parts = ax5.violinplot(violin_data, positions=[1, 2], showmeans=True, showextrema=True)
        
        # Color the violin plots
        for i, (pc, split) in enumerate(zip(parts['bodies'], ['val', 'test'])):
            pc.set_facecolor(colors[split])
            pc.set_alpha(0.7)
        
        ax5.set_xticks([1, 2])
        ax5.set_xticklabels(violin_labels)
        ax5.set_title(f'Violin Plot for {pheno.upper()} (Validation vs Test)', 
                     fontweight='bold', pad=20)
        ax5.set_ylabel('Residual Value', fontweight='bold')
        ax5.grid(True, alpha=0.3, axis='y')
        plt.tight_layout()
        plt.show()
        
        # 6. Scatter Plot: Predictions vs True Values
        fig6, ax6 = plt.subplots(figsize=(8, 6))
        
        for split in ['val', 'test']:
            df = datasets[split]
            if not df.empty and pred_col in df.columns and label_col in df.columns:
                predictions = df[pred_col]
                true_values = df[label_col]
                
                ax6.scatter(true_values, predictions, color=colors[split], 
                           alpha=0.6, s=20, label=split_names[split], edgecolors='white', linewidth=0.3)
        
        # Add perfect prediction line
        all_true = []
        all_pred = []
        for split in ['val', 'test']:
            df = datasets[split]
            if not df.empty and pred_col in df.columns and label_col in df.columns:
                all_true.extend(df[label_col].tolist())
                all_pred.extend(df[pred_col].tolist())
        
        if all_true and all_pred:
            min_val = min(min(all_true), min(all_pred))
            max_val = max(max(all_true), max(all_pred))
            ax6.plot([min_val, max_val], [min_val, max_val], 'k--', 
                    alpha=0.8, linewidth=2, label='Perfect Prediction')
        
        ax6.set_title(f'Predictions vs True Values for {pheno.upper()} (Validation vs Test)', 
                     fontweight='bold', pad=20)
        ax6.set_xlabel('True Values', fontweight='bold')
        ax6.set_ylabel('Predicted Values', fontweight='bold')
        ax6.legend(frameon=True, fancybox=True, shadow=True)
        ax6.grid(True, alpha=0.3)
        plt.tight_layout()
        plt.show()
        
        # Statistical Analysis Results
        print(f"\n--- Statistical Analysis Results ---")
        
        # Basic descriptive statistics
        print("\nDescriptive Statistics:")
        print("-" * 50)
        print(f"{'Metric':<20} {'Validation':<15} {'Test':<15}")
        print("-" * 50)
        
        metrics = [
            ('Sample Size', lambda x: len(x)),
            ('Mean', lambda x: x.mean()),
            ('Std Dev', lambda x: x.std()),
            ('Variance', lambda x: x.var()),
            ('Skewness', lambda x: scipy_stats.skew(x)),
            ('Kurtosis', lambda x: scipy_stats.kurtosis(x)),
            ('Minimum', lambda x: x.min()),
            ('Maximum', lambda x: x.max()),
            ('Median', lambda x: x.median()),
            ('MAE', lambda x: x.abs().mean())
        ]
        
        for metric_name, metric_func in metrics:
            val_val = metric_func(residuals_data['val'])
            test_val = metric_func(residuals_data['test'])
            if metric_name == 'Sample Size':
                print(f"{metric_name:<20} {val_val:<15.0f} {test_val:<15.0f}")
            else:
                print(f"{metric_name:<20} {val_val:<15.6f} {test_val:<15.6f}")
        
        # Normality Tests
        print("\n\nNormality Tests:")
        print("-" * 50)
        
        for split, residuals in residuals_data.items():
            print(f"\n{split_names[split]} Set:")
            
            # Shapiro-Wilk Test
            if len(residuals) >= 3:
                shapiro_stat, shapiro_p = scipy_stats.shapiro(residuals)
                print(f"  Shapiro-Wilk Test:")
                print(f"    Statistic: {shapiro_stat:.6f}")
                print(f"    p-value: {shapiro_p:.6f}")
                print(f"    Result: {'Normal distribution' if shapiro_p > 0.05 else 'Non-normal distribution'}")
            
            # Kolmogorov-Smirnov Test
            residuals_normalized = (residuals - residuals.mean()) / residuals.std()
            ks_stat, ks_p = scipy_stats.kstest(residuals_normalized, 'norm')
            print(f"  Kolmogorov-Smirnov Test:")
            print(f"    Statistic: {ks_stat:.6f}")
            print(f"    p-value: {ks_p:.6f}")
            print(f"    Result: {'Normal distribution' if ks_p > 0.05 else 'Non-normal distribution'}")
        
        # Two-sample comparison tests
        val_residuals = residuals_data['val']
        test_residuals = residuals_data['test']
        
        print("\n\nTwo-Sample Comparison Tests:")
        print("-" * 50)
        
        # Mann-Whitney U Test (non-parametric)
        mw_stat, mw_p = scipy_stats.mannwhitneyu(val_residuals, test_residuals, 
                                                 alternative='two-sided')
        print(f"Mann-Whitney U Test:")
        print(f"  Statistic: {mw_stat:.6f}")
        print(f"  p-value: {mw_p:.6f}")
        print(f"  Result: {'No significant difference' if mw_p > 0.05 else 'Significant difference'} in distributions")
        
        # Levene's Test for Equal Variances
        levene_stat, levene_p = scipy_stats.levene(val_residuals, test_residuals)
        print(f"\nLevene's Test for Equal Variances:")
        print(f"  Statistic: {levene_stat:.6f}")
        print(f"  p-value: {levene_p:.6f}")
        print(f"  Result: {'Equal variances' if levene_p > 0.05 else 'Unequal variances'}")
        
        # t-test (parametric)
        if levene_p > 0.05:
            t_stat, t_p = scipy_stats.ttest_ind(val_residuals, test_residuals, equal_var=True)
            test_name = "Student's t-test"
        else:
            t_stat, t_p = scipy_stats.ttest_ind(val_residuals, test_residuals, equal_var=False)
            test_name = "Welch's t-test"
        
        print(f"\n{test_name}:")
        print(f"  Statistic: {t_stat:.6f}")
        print(f"  p-value: {t_p:.6f}")
        print(f"  Result: {'No significant difference' if t_p > 0.05 else 'Significant difference'} in means")
        
        # Effect Size (Cohen's d)
        pooled_std = np.sqrt(((len(val_residuals) - 1) * val_residuals.var() + 
                             (len(test_residuals) - 1) * test_residuals.var()) / 
                            (len(val_residuals) + len(test_residuals) - 2))
        cohens_d = (val_residuals.mean() - test_residuals.mean()) / pooled_std
        
        print(f"\nEffect Size (Cohen's d): {cohens_d:.6f}")
        
        if abs(cohens_d) < 0.2:
            effect_size = "Small"
        elif abs(cohens_d) < 0.8:
            effect_size = "Medium"
        else:
            effect_size = "Large"
        
        print(f"Effect Size Interpretation: {effect_size}")
        print("=" * 60)

def plot_combined_val_test_boxplot(datasets, phenotypes):
    """
    Create combined box plot for all phenotypes comparing validation and test sets
    """
    # Collect residuals data
    residuals_data = {}
    
    for split in ['val', 'test']:
        residuals_data[split] = {}
        df = datasets[split]
        
        if df.empty:
            continue
            
        for pheno in phenotypes:
            pred_col = f"{pheno}_pred"
            label_col = f"{pheno}_label"
            
            if pred_col in df.columns and label_col in df.columns:
                residuals = df[pred_col] - df[label_col]
                if not residuals.empty and residuals.nunique() >= 2:
                    residuals_data[split][pheno] = residuals
    
    # Create the combined box plot
    fig, ax = plt.subplots(figsize=(12, 6))
    
    # Prepare data for box plot
    box_data = []
    box_labels = []
    colors_list = []
    positions = []
    
    pos = 1
    for pheno in phenotypes:
        # Check if both val and test have data for this phenotype
        if pheno in residuals_data.get('val', {}) and pheno in residuals_data.get('test', {}):
            # Add validation data
            box_data.append(residuals_data['val'][pheno])
            box_labels.append(f'{pheno.upper()}\nValidation')
            colors_list.append(colors['val'])
            positions.append(pos)
            
            # Add test data
            box_data.append(residuals_data['test'][pheno])
            box_labels.append(f'{pheno.upper()}\nTest')
            colors_list.append(colors['test'])
            positions.append(pos + 1)
            
            pos += 3  # Leave space between phenotype groups
    
    if box_data:
        # Create box plot with custom positions
        box_plot = ax.boxplot(box_data, positions=positions, labels=box_labels, 
                             patch_artist=True,
                             boxprops=dict(linewidth=1.5),
                             medianprops=dict(linewidth=2, color='black'),
                             whiskerprops=dict(linewidth=1.5),
                             capprops=dict(linewidth=1.5),
                             widths=0.6)
        
        # Set colors
        for patch, color in zip(box_plot['boxes'], colors_list):
            patch.set_facecolor(color)
            patch.set_alpha(0.7)
        
        ax.set_title('Pre-trained Model: Validation vs Test Set Comparison', 
                    fontweight='bold', pad=20)
        ax.set_ylabel('Residual Value', fontweight='bold')
        ax.grid(True, alpha=0.3, axis='y')
        
        # Rotate x-axis labels for better readability
        ax.tick_params(axis='x', rotation=45)
        
        # Add legend
        from matplotlib.patches import Patch
        legend_elements = [Patch(facecolor=colors['val'], alpha=0.7, label='Validation'),
                          Patch(facecolor=colors['test'], alpha=0.7, label='Test')]
        ax.legend(handles=legend_elements, loc='upper right', frameon=True, 
                 fancybox=True, shadow=True)
        
    else:
        ax.text(0.5, 0.5, 'No data available for comparison', 
               transform=ax.transAxes, ha='center', va='center', fontsize=14)
        ax.set_title('Pre-trained Model - No Data Available', fontweight='bold')
    
    plt.tight_layout()
    plt.show()

# Execute the comprehensive analysis
if phenotypes and not all(df.empty for df in pretrained_datasets.values()):
    print(f"\n{'='*80}")
    print("COMPREHENSIVE ANALYSIS: PRE-TRAINED MODEL VALIDATION vs TEST")
    print(f"{'='*80}")
    
    # Generate comprehensive comparison plots for each phenotype
    compare_val_test_analysis_publication(pretrained_datasets, phenotypes)
    
    # Generate the combined box plot for all phenotypes
    print(f"\n{'='*80}")
    print("COMBINED BOX PLOT FOR ALL PHENOTYPES")
    print(f"{'='*80}")
    plot_combined_val_test_boxplot(pretrained_datasets, phenotypes)
    
else:
    print("No valid data found for analysis")
```

```python
def plot_combined_all_phenotypes_boxplot(datasets, phenotypes):
    """
    Create combined box plot for all phenotypes comparing validation and test sets on one figure
    """
    # Collect residuals data
    residuals_data = {}
    
    for split in ['val', 'test']:
        residuals_data[split] = {}
        df = datasets[split]
        
        if df.empty:
            continue
            
        for pheno in phenotypes:
            pred_col = f"{pheno}_pred"
            label_col = f"{pheno}_label"
            
            if pred_col in df.columns and label_col in df.columns:
                residuals = df[pred_col] - df[label_col]
                if not residuals.empty and residuals.nunique() >= 2:
                    residuals_data[split][pheno] = residuals
    
    # Create the combined box plot
    fig, ax = plt.subplots(figsize=(12, 6))
    
    # Prepare data for box plot
    box_data = []
    box_labels = []
    colors_list = []
    positions = []
    
    pos = 1
    for pheno in phenotypes:
        # Check if both val and test have data for this phenotype
        if pheno in residuals_data.get('val', {}) and pheno in residuals_data.get('test', {}):
            # Add validation data
            box_data.append(residuals_data['val'][pheno])
            box_labels.append(f'{pheno.upper()}\nValidation')
            colors_list.append(colors['val'])
            positions.append(pos)
            
            # Add test data
            box_data.append(residuals_data['test'][pheno])
            box_labels.append(f'{pheno.upper()}\nTest')
            colors_list.append(colors['test'])
            positions.append(pos + 1)
            
            pos += 3  # Leave space between phenotype groups
    
    if box_data:
        # Create box plot with custom positions
        box_plot = ax.boxplot(box_data, positions=positions, labels=box_labels, 
                             patch_artist=True,
                             boxprops=dict(linewidth=1.5),
                             medianprops=dict(linewidth=2, color='black'),
                             whiskerprops=dict(linewidth=1.5),
                             capprops=dict(linewidth=1.5),
                             widths=0.6)
        
        # Set colors
        for patch, color in zip(box_plot['boxes'], colors_list):
            patch.set_facecolor(color)
            patch.set_alpha(0.7)
        
        ax.set_title('Box Plot: All Phenotypes Validation vs Test Comparison', 
                    fontweight='bold', pad=20)
        ax.set_ylabel('Residual Value', fontweight='bold')
        ax.grid(True, alpha=0.3, axis='y')
        
        # Rotate x-axis labels for better readability
        ax.tick_params(axis='x', rotation=45)
        
        # Add legend
        from matplotlib.patches import Patch
        legend_elements = [Patch(facecolor=colors['val'], alpha=0.7, label='Validation'),
                          Patch(facecolor=colors['test'], alpha=0.7, label='Test')]
        ax.legend(handles=legend_elements, loc='upper right', frameon=True, 
                 fancybox=True, shadow=True)
        
    else:
        ax.text(0.5, 0.5, 'No data available for comparison', 
               transform=ax.transAxes, ha='center', va='center', fontsize=14)
        ax.set_title('Box Plot - No Data Available', fontweight='bold')
    
    plt.tight_layout()
    plt.show()

def plot_combined_all_phenotypes_violin(datasets, phenotypes):
    """
    Create combined violin plot for all phenotypes comparing validation and test sets on one figure
    """
    # Collect residuals data
    residuals_data = {}
    
    for split in ['val', 'test']:
        residuals_data[split] = {}
        df = datasets[split]
        
        if df.empty:
            continue
            
        for pheno in phenotypes:
            pred_col = f"{pheno}_pred"
            label_col = f"{pheno}_label"
            
            if pred_col in df.columns and label_col in df.columns:
                residuals = df[pred_col] - df[label_col]
                if not residuals.empty and residuals.nunique() >= 2:
                    residuals_data[split][pheno] = residuals
    
    # Create the combined violin plot
    fig, ax = plt.subplots(figsize=(12, 6))
    
    # Prepare data for violin plot
    violin_data = []
    violin_labels = []
    positions = []
    
    pos = 1
    valid_phenotypes = []
    for pheno in phenotypes:
        # Check if both val and test have data for this phenotype
        if pheno in residuals_data.get('val', {}) and pheno in residuals_data.get('test', {}):
            # Add validation data
            violin_data.append(residuals_data['val'][pheno])
            violin_labels.append(f'{pheno.upper()}\nValidation')
            positions.append(pos)
            
            # Add test data
            violin_data.append(residuals_data['test'][pheno])
            violin_labels.append(f'{pheno.upper()}\nTest')
            positions.append(pos + 1)
            
            valid_phenotypes.append(pheno)
            pos += 3  # Leave space between phenotype groups
    
    if violin_data:
        # Create violin plot with custom positions
        parts = ax.violinplot(violin_data, positions=positions, showmeans=True, 
                             showextrema=True, showmedians=True)
        
        # Color the violin plots
        color_sequence = []
        for pheno in valid_phenotypes:
            color_sequence.extend([colors['val'], colors['test']])
        
        for pc, color in zip(parts['bodies'], color_sequence):
            pc.set_facecolor(color)
            pc.set_alpha(0.7)
            pc.set_edgecolor('black')
            pc.set_linewidth(1)
        
        # Customize other elements
        for partname in ('cbars', 'cmins', 'cmaxes', 'cmedians', 'cmeans'):
            if partname in parts:
                vp = parts[partname]
                vp.set_edgecolor('black')
                vp.set_linewidth(1.5)
        
        # Set labels
        ax.set_xticks(positions)
        ax.set_xticklabels(violin_labels)
        
        ax.set_title('Violin Plot: All Phenotypes Validation vs Test Comparison', 
                    fontweight='bold', pad=20)
        ax.set_ylabel('Residual Value', fontweight='bold')
        ax.grid(True, alpha=0.3, axis='y')
        
        # Rotate x-axis labels for better readability
        ax.tick_params(axis='x', rotation=45)
        
        # Add legend
        from matplotlib.patches import Patch
        legend_elements = [Patch(facecolor=colors['val'], alpha=0.7, label='Validation'),
                          Patch(facecolor=colors['test'], alpha=0.7, label='Test')]
        ax.legend(handles=legend_elements, loc='upper right', frameon=True, 
                 fancybox=True, shadow=True)
        
    else:
        ax.text(0.5, 0.5, 'No data available for comparison', 
               transform=ax.transAxes, ha='center', va='center', fontsize=14)
        ax.set_title('Violin Plot - No Data Available', fontweight='bold')
    
    plt.tight_layout()
    plt.show()

# 在原有代码的执行部分添加这两个新函数的调用
if phenotypes and not all(df.empty for df in pretrained_datasets.values()):
    print(f"\n{'='*80}")
    print("COMPREHENSIVE ANALYSIS: PRE-TRAINED MODEL VALIDATION vs TEST")
    print(f"{'='*80}")
    
    # Generate comprehensive comparison plots for each phenotype
    compare_val_test_analysis_publication(pretrained_datasets, phenotypes)
    
    # Generate the combined plots for all phenotypes
    print(f"\n{'='*80}")
    print("COMBINED BOX PLOT FOR ALL PHENOTYPES")
    print(f"{'='*80}")
    plot_combined_all_phenotypes_boxplot(pretrained_datasets, phenotypes)
    
    print(f"\n{'='*80}")
    print("COMBINED VIOLIN PLOT FOR ALL PHENOTYPES")
    print(f"{'='*80}")
    plot_combined_all_phenotypes_violin(pretrained_datasets, phenotypes)
    
else:
    print("No valid data found for analysis")
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
import seaborn as sns
import os

# 读取数据
# 假设数据存储在项目根目录的data文件夹中，您可能需要调整路径
# 使用制表符分隔的格式读取文件
file_path = '/media/marxin/softs/workspace/Whisper_of_DNA_pl/data/maize1404_genotype/A_cubic_1404_mic_qc_MIC.bimmic'  # 调整为您的实际文件路径
data = pd.read_csv(file_path, sep='\t')

# 检查数据的基本信息
print("数据形状:", data.shape)
print("数据列:", data.columns.tolist())
print("前几行数据:")
data.head()
```

```python
# 提取所有表型名称（格式为xxx_mic的列）
phenotypes = [col.replace('_mic', '') for col in data.columns if col.endswith('_mic')]
print(f"发现{len(phenotypes)}个表型:", phenotypes)

# 定义绘制曼哈顿图的函数，使用更加温暖舒适的颜色
def manhattan_plot(data, phenotype, highlight_threshold=0.01):
    """
    绘制指定表型的曼哈顿图，使用温暖舒适的颜色方案
    
    参数:
    - data: 包含SNP数据的DataFrame
    - phenotype: 要绘制的表型名称
    - highlight_threshold: 高亮显示的阈值
    """
    # 准备数据
    plot_data = data.copy()
    # 将染色体转换为数字以便排序
    plot_data['chr'] = plot_data['chr'].astype(str)
    plot_data['chr'] = pd.to_numeric(plot_data['chr'].str.replace('chr', ''), errors='coerce')
    
    # 按染色体和位置排序
    plot_data = plot_data.sort_values(['chr', 'position'])
    
    # 计算x轴的位置
    # 为每个染色体创建一个累计位置
    chromosomes = plot_data['chr'].unique()
    chromosomes.sort()
    
    # 获取每条染色体的最大位置
    max_positions = {chrom: plot_data[plot_data['chr'] == chrom]['position'].max() for chrom in chromosomes}
    
    # 计算每个SNP的x轴位置
    plot_data['cumulative_pos'] = 0
    cum_pos = 0
    
    for i, chrom in enumerate(chromosomes):
        plot_data.loc[plot_data['chr'] == chrom, 'cumulative_pos'] = plot_data.loc[plot_data['chr'] == chrom, 'position'] + cum_pos
        cum_pos += max_positions[chrom] + 5000000  # 在染色体之间添加一定的间隔
    
    # 计算每个染色体的中点位置，用于x轴标记
    chrom_labels = {}
    for chrom in chromosomes:
        chrom_data = plot_data[plot_data['chr'] == chrom]
        chrom_labels[chrom] = (chrom_data['cumulative_pos'].min() + chrom_data['cumulative_pos'].max()) / 2
    
    # 使用-log10(mic)作为y轴值
    mic_col = f"{phenotype}_mic"
    plot_data['-log10(mic)'] = -np.log10(plot_data[mic_col])
    
    # 设置绘图样式
    plt.figure(figsize=(14, 7))
    
    # 设置温暖舒适的染色体颜色 - 使用更多的颜色区分染色体
    # 温暖的色调：橙色、珊瑚色、浅棕色、桃色等
    warm_colors = [
        '#E67E22', '#F39C12', '#D35400', '#F1C40F',  # 橙色系
        '#F5B041', '#E59866', '#FAD7A0', '#EDBB99',  # 黄棕色系
        '#DC7633', '#EB984E', '#F0B27A', '#FDEBD0',  # 桃色系
        '#CA6F1E', '#BA4A00', '#AF601A', '#B9770E'   # 红棕色系
    ]
    
    # 绘制点
    for i, chrom in enumerate(chromosomes):
        chrom_data = plot_data[plot_data['chr'] == chrom]
        plt.scatter(
            chrom_data['cumulative_pos'], 
            chrom_data['-log10(mic)'],
            c=warm_colors[i % len(warm_colors)],
            alpha=0.8,
            s=18,
            label=f"Chr {chrom}"
        )
    
    # 高亮显示显著的SNP
    sig_snps = plot_data[plot_data[mic_col] < highlight_threshold]
    if not sig_snps.empty:
        plt.scatter(
            sig_snps['cumulative_pos'],
            sig_snps['-log10(mic)'],
            c='#E74C3C',  # 更温暖的红色
            alpha=0.9,
            s=35,
            edgecolors='black',
            linewidths=0.5
        )
    
    # 设置x轴标签
    plt.xticks([chrom_labels[chrom] for chrom in chromosomes], [f"Chr {chrom}" for chrom in chromosomes])
    plt.xlim(0, plot_data['cumulative_pos'].max() + 5000000)
    
    # 添加标题和标签
    plt.title(f"Manhattan Plot for {phenotype}", fontsize=18)
    plt.xlabel("Chromosome", fontsize=14)
    plt.ylabel("-log10(Weights)", fontsize=14)
    
    # 添加图例 - 只添加显著SNP图例
    if not sig_snps.empty:
        red_patch = mpatches.Patch(color='#E74C3C', label=f'显著SNP (MIC < {highlight_threshold})')
        plt.legend(handles=[red_patch], loc='upper right')
    
    # 美化图表
    plt.grid(axis='y', linestyle='--', alpha=0.3)
    plt.tight_layout()
    return plt

# 绘制第一个表型的曼哈顿图作为示例
manhattan_plot(data, phenotypes[2])
plt.show()
manhattan_plot(data, phenotypes[3])
plt.show()
manhattan_plot(data, phenotypes[4])
plt.show()
```

```python
# 创建多个表型的曼哈顿图
# 可以选择几个表型进行展示
selected_phenotypes = phenotypes[:4]  # 选择前4个表型，您可以根据需要调整

plt.figure(figsize=(15, 12))
for i, phenotype in enumerate(selected_phenotypes):
    plt.subplot(2, 2, i+1)
    manhattan_plot(data, phenotype)
    plt.tight_layout()

plt.suptitle("Manhattan Plots for Selected Phenotypes", fontsize=16, y=1.02)
plt.tight_layout()
plt.show()
```
