```python
import pandas as pd
import matplotlib.pyplot as plt
import os
import numpy as np
from scipy.stats import pearsonr
import scipy.stats as stats
import seaborn as sns
from scipy import stats as scipy_stats

# Set publication-ready style
plt.style.use('seaborn-v0_8-whitegrid')
sns.set_palette("husl")

# Configure matplotlib for publication quality
plt.rcParams.update({
    'font.size': 12,
    'font.family': 'serif',
    'font.serif': ['Times New Roman'],
    'axes.labelsize': 14,
    'axes.titlesize': 16,
    'xtick.labelsize': 12,
    'ytick.labelsize': 12,
    'legend.fontsize': 12,
    'figure.titlesize': 18,
    'lines.linewidth': 2,
    'axes.linewidth': 1.2,
    'grid.alpha': 0.3,
    'axes.spines.top': False,
    'axes.spines.right': False
})

# Define file paths
log_dirfinetunes = '../logs0/DNAWhisperfinetune/run0_blockembedding_f3p_30k/'  # Pre-trained model
log_dir = '../logs0/DNAWhisper/run16_blockembedding_f3p_30k/'  # Non-pre-trained model

# Load datasets for both models
datasets = {}
model_names = {'pretrained': 'Pre-trained', 'no_pretrain': 'Non-pre-trained'}
colors = {'pretrained': '#2E86AB', 'no_pretrain': '#A23B72'}  # Professional color scheme

for name, path in [('pretrained', log_dirfinetunes), ('no_pretrain', log_dir)]:
    train_file = os.path.join(path, 'train_predictions.csv')
    val_file = os.path.join(path, 'val_predictions.csv')
    test_file = os.path.join(path, 'test_predictions.csv')
    
    try:
        datasets[name] = {
            'train': pd.read_csv(train_file),
            'val': pd.read_csv(val_file),
            'test': pd.read_csv(test_file)
        }
        print(f"Successfully loaded {model_names[name]} datasets")
    except FileNotFoundError as e:
        print(f"Error loading {model_names[name]} files: {e}")
        datasets[name] = {'train': pd.DataFrame(), 'val': pd.DataFrame(), 'test': pd.DataFrame()}

# Get phenotypes from the first available dataset
phenotypes = []
for dataset_name, dataset_dict in datasets.items():
    for split_name, df in dataset_dict.items():
        if not df.empty:
            phenotypes = sorted(list(set([col.split('_')[0] for col in df.columns if '_' in col])))
            break
    if phenotypes:
        break

print(f"Found phenotypes: {phenotypes}")
```

```python
def compare_models_analysis_publication(datasets, phenotypes, split='train'):
    """
    Publication-ready comparison analysis of pre-trained and non-pre-trained models
    Each plot is generated separately for easy integration into papers
    """
    
    for pheno in phenotypes:
        pred_col = f"{pheno}_pred"
        label_col = f"{pheno}_label"
        
        # Check if both models have data for this phenotype
        residuals_data = {}
        for model_name in ['pretrained', 'no_pretrain']:
            df = datasets[model_name][split]
            if not df.empty and pred_col in df.columns and label_col in df.columns:
                residuals = df[pred_col] - df[label_col]
                if not residuals.empty and residuals.nunique() >= 2:
                    residuals_data[model_name] = residuals
        
        if len(residuals_data) < 2:
            print(f"Skipping phenotype {pheno} for {split} set: Insufficient data")
            continue
            
        print(f"\n=== Phenotype {pheno.upper()} Comparison Analysis ({split.upper()} set) ===")
        
        # 1. Q-Q Plot Comparison
        fig1, ax1 = plt.subplots(figsize=(7, 5))
        
        # Get theoretical quantiles only once
        sample_size = min(len(residuals_data['pretrained']), len(residuals_data['no_pretrain']))
        theoretical_quantiles = scipy_stats.norm.ppf(np.linspace(0.01, 0.99, sample_size))
        
        # Plot sample quantiles for both models
        for model_name, residuals in residuals_data.items():
            sample_quantiles = np.sort(residuals)
            # Interpolate to match theoretical quantiles length
            if len(sample_quantiles) != len(theoretical_quantiles):
                x_interp = np.linspace(0, 1, len(sample_quantiles))
                x_new = np.linspace(0, 1, len(theoretical_quantiles))
                sample_quantiles = np.interp(x_new, x_interp, sample_quantiles)
            
            ax1.scatter(theoretical_quantiles, sample_quantiles, 
                       color=colors[model_name], alpha=0.7, s=20,
                       label=model_names[model_name], edgecolors='white', linewidth=0.5)
        
        # Add single reference line for normal distribution
        min_val = min(theoretical_quantiles.min(), 
                     min([np.sort(res)[0] for res in residuals_data.values()]))
        max_val = max(theoretical_quantiles.max(), 
                     max([np.sort(res)[-1] for res in residuals_data.values()]))
        ax1.plot([min_val, max_val], [min_val, max_val], 'k--', 
                alpha=0.8, linewidth=2, label='Normal Distribution')
        
        ax1.set_title(f'Q-Q Plot for {pheno.upper()} ({split.upper()})', fontweight='bold', pad=20)
        ax1.set_xlabel('Theoretical Quantiles', fontweight='bold')
        ax1.set_ylabel('Sample Quantiles', fontweight='bold')
        ax1.legend(frameon=True, fancybox=True, shadow=True)
        ax1.grid(True, alpha=0.3)
        plt.tight_layout()
        plt.show()
        
        # 2. Histogram Comparison with KDE
        fig2, ax2 = plt.subplots(figsize=(7, 5))
        
        for model_name, residuals in residuals_data.items():
            ax2.hist(residuals, bins=25, alpha=0.6, color=colors[model_name], 
                    label=model_names[model_name], density=True, edgecolor='white', linewidth=0.7)
            
            # Add KDE overlay
            kde_x = np.linspace(residuals.min(), residuals.max(), 100)
            kde = scipy_stats.gaussian_kde(residuals)
            ax2.plot(kde_x, kde(kde_x), color=colors[model_name], linewidth=2.5, alpha=0.8)
        
        ax2.set_title(f'Residual Distribution for {pheno.upper()} ({split.upper()})', fontweight='bold', pad=20)
        ax2.set_xlabel('Residual Value', fontweight='bold')
        ax2.set_ylabel('Density', fontweight='bold')
        ax2.legend(frameon=True, fancybox=True, shadow=True)
        ax2.grid(True, alpha=0.3)
        plt.tight_layout()
        plt.show()
        
        # 3. Box Plot Comparison
        fig3, ax3 = plt.subplots(figsize=(7, 5))
        
        box_data = [residuals_data['pretrained'], residuals_data['no_pretrain']]
        box_labels = ['Pre-trained', 'Non-pre-trained']
        
        box_plot = ax3.boxplot(box_data, labels=box_labels, patch_artist=True,
                              boxprops=dict(linewidth=1.5),
                              medianprops=dict(linewidth=2, color='black'),
                              whiskerprops=dict(linewidth=1.5),
                              capprops=dict(linewidth=1.5))
        
        # Set box colors
        box_plot['boxes'][0].set_facecolor(colors['pretrained'])
        box_plot['boxes'][0].set_alpha(0.7)
        box_plot['boxes'][1].set_facecolor(colors['no_pretrain'])
        box_plot['boxes'][1].set_alpha(0.7)
        
        ax3.set_title(f'Box Plot Comparison for {pheno.upper()} ({split.upper()})', fontweight='bold', pad=20)
        ax3.set_ylabel('Residual Value', fontweight='bold')
        ax3.grid(True, alpha=0.3, axis='y')
        plt.tight_layout()
        plt.show()
        
        # 4. Kernel Density Estimation Comparison
        fig4, ax4 = plt.subplots(figsize=(7, 5))
        
        for model_name, residuals in residuals_data.items():
            sns.kdeplot(data=residuals, ax=ax4, color=colors[model_name], 
                       label=model_names[model_name], linewidth=2.5, alpha=0.8)
        
        ax4.set_title(f'Kernel Density Estimation for {pheno.upper()} ({split.upper()})', fontweight='bold', pad=20)
        ax4.set_xlabel('Residual Value', fontweight='bold')
        ax4.set_ylabel('Density', fontweight='bold')
        ax4.legend(frameon=True, fancybox=True, shadow=True)
        ax4.grid(True, alpha=0.3)
        plt.tight_layout()
        plt.show()
        
        # Statistical Analysis Results
        print(f"\n--- Statistical Analysis Results for {split.upper()} Set ---")
        
        # Basic descriptive statistics
        print("\nDescriptive Statistics:")
        print("-" * 60)
        print(f"{'Metric':<20} {'Pre-trained':<15} {'Non-pre-trained':<15}")
        print("-" * 60)
        
        metrics = [
            ('Sample Size', lambda x: len(x)),
            ('Mean', lambda x: x.mean()),
            ('Std Dev', lambda x: x.std()),
            ('Variance', lambda x: x.var()),
            ('Skewness', lambda x: scipy_stats.skew(x)),
            ('Kurtosis', lambda x: scipy_stats.kurtosis(x)),
            ('Minimum', lambda x: x.min()),
            ('Maximum', lambda x: x.max()),
            ('Median', lambda x: x.median()),
            ('MAE', lambda x: x.abs().mean())
        ]
        
        for metric_name, metric_func in metrics:
            pre_val = metric_func(residuals_data['pretrained'])
            no_pre_val = metric_func(residuals_data['no_pretrain'])
            if metric_name == 'Sample Size':
                print(f"{metric_name:<20} {pre_val:<15.0f} {no_pre_val:<15.0f}")
            else:
                print(f"{metric_name:<20} {pre_val:<15.6f} {no_pre_val:<15.6f}")
        
        # Normality Tests
        print("\n\nNormality Tests:")
        print("-" * 60)
        
        for model_name, residuals in residuals_data.items():
            print(f"\n{model_names[model_name]} Model:")
            
            # Shapiro-Wilk Test
            if len(residuals) >= 3:
                shapiro_stat, shapiro_p = scipy_stats.shapiro(residuals)
                print(f"  Shapiro-Wilk Test:")
                print(f"    Statistic: {shapiro_stat:.6f}")
                print(f"    p-value: {shapiro_p:.6f}")
                print(f"    Result: {'Normal distribution' if shapiro_p > 0.05 else 'Non-normal distribution'}")
            
            # Kolmogorov-Smirnov Test
            residuals_normalized = (residuals - residuals.mean()) / residuals.std()
            ks_stat, ks_p = scipy_stats.kstest(residuals_normalized, 'norm')
            print(f"  Kolmogorov-Smirnov Test:")
            print(f"    Statistic: {ks_stat:.6f}")
            print(f"    p-value: {ks_p:.6f}")
            print(f"    Result: {'Normal distribution' if ks_p > 0.05 else 'Non-normal distribution'}")
        
        # Two-sample comparison tests
        pretrained_residuals = residuals_data['pretrained']
        no_pretrain_residuals = residuals_data['no_pretrain']
        
        print("\n\nTwo-Sample Comparison Tests:")
        print("-" * 60)
        
        # Mann-Whitney U Test (non-parametric)
        mw_stat, mw_p = scipy_stats.mannwhitneyu(pretrained_residuals, no_pretrain_residuals, 
                                                 alternative='two-sided')
        print(f"Mann-Whitney U Test:")
        print(f"  Statistic: {mw_stat:.6f}")
        print(f"  p-value: {mw_p:.6f}")
        print(f"  Result: {'No significant difference' if mw_p > 0.05 else 'Significant difference'} in distributions")
        
        # Levene's Test for Equal Variances
        levene_stat, levene_p = scipy_stats.levene(pretrained_residuals, no_pretrain_residuals)
        print(f"\nLevene's Test for Equal Variances:")
        print(f"  Statistic: {levene_stat:.6f}")
        print(f"  p-value: {levene_p:.6f}")
        print(f"  Result: {'Equal variances' if levene_p > 0.05 else 'Unequal variances'}")
        
        # t-test (parametric)
        if levene_p > 0.05:
            t_stat, t_p = scipy_stats.ttest_ind(pretrained_residuals, no_pretrain_residuals, 
                                               equal_var=True)
            test_name = "Student's t-test"
        else:
            t_stat, t_p = scipy_stats.ttest_ind(pretrained_residuals, no_pretrain_residuals, 
                                               equal_var=False)
            test_name = "Welch's t-test"
        
        print(f"\n{test_name}:")
        print(f"  Statistic: {t_stat:.6f}")
        print(f"  p-value: {t_p:.6f}")
        print(f"  Result: {'No significant difference' if t_p > 0.05 else 'Significant difference'} in means")
        
        # Effect Size (Cohen's d)
        pooled_std = np.sqrt(((len(pretrained_residuals) - 1) * pretrained_residuals.var() + 
                             (len(no_pretrain_residuals) - 1) * no_pretrain_residuals.var()) / 
                            (len(pretrained_residuals) + len(no_pretrain_residuals) - 2))
        cohens_d = (pretrained_residuals.mean() - no_pretrain_residuals.mean()) / pooled_std
        
        print(f"\nEffect Size (Cohen's d): {cohens_d:.6f}")
        
        if abs(cohens_d) < 0.2:
            effect_size = "Small"
        elif abs(cohens_d) < 0.8:
            effect_size = "Medium"
        else:
            effect_size = "Large"
        
        print(f"Effect Size Interpretation: {effect_size}")
        print("=" * 60)

# Execute comparison analysis for all splits
splits = ['train', 'val', 'test']

for split in splits:
    print(f"\n{'='*80}")
    print(f"ANALYZING {split.upper()} DATASET")
    print(f"{'='*80}")
    
    # Check if both models have data for this split
    has_data = False
    for model_name in ['pretrained', 'no_pretrain']:
        if not datasets[model_name][split].empty:
            has_data = True
            break
    
    if not has_data:
        print(f"No data available for {split} set")
        continue
    
    compare_models_analysis_publication(datasets, phenotypes, split=split)
```

```python
def plot_combined_boxplots(datasets, phenotypes, splits=['val', 'test']):
    """
    Create combined box plots for validation and test sets across all phenotypes
    """
    # Collect all residuals data
    all_residuals_data = {}
    
    for split in splits:
        all_residuals_data[split] = {}
        for pheno in phenotypes:
            pred_col = f"{pheno}_pred"
            label_col = f"{pheno}_label"
            
            residuals_data = {}
            for model_name in ['pretrained', 'no_pretrain']:
                df = datasets[model_name][split]
                if not df.empty and pred_col in df.columns and label_col in df.columns:
                    residuals = df[pred_col] - df[label_col]
                    if not residuals.empty and residuals.nunique() >= 2:
                        residuals_data[model_name] = residuals
            
            if len(residuals_data) == 2:  # Both models have data
                all_residuals_data[split][pheno] = residuals_data
    
    # Create the combined box plot
    fig, axes = plt.subplots(1, len(splits), figsize=(7*len(splits), 6))
    if len(splits) == 1:
        axes = [axes]
    
    for idx, split in enumerate(splits):
        ax = axes[idx]
        
        # Prepare data for box plot
        box_data = []
        box_labels = []
        colors_list = []
        
        for pheno in phenotypes:
            if pheno in all_residuals_data[split]:
                # Add pre-trained data
                box_data.append(all_residuals_data[split][pheno]['pretrained'])
                box_labels.append(f'{pheno.upper()}\nPre-trained')
                colors_list.append(colors['pretrained'])
                
                # Add non-pre-trained data
                box_data.append(all_residuals_data[split][pheno]['no_pretrain'])
                box_labels.append(f'{pheno.upper()}\nNon-pre-trained')
                colors_list.append(colors['no_pretrain'])
        
        if box_data:
            # Create box plot
            box_plot = ax.boxplot(box_data, labels=box_labels, patch_artist=True,
                                 boxprops=dict(linewidth=1.5),
                                 medianprops=dict(linewidth=2, color='black'),
                                 whiskerprops=dict(linewidth=1.5),
                                 capprops=dict(linewidth=1.5))
            
            # Set colors
            for patch, color in zip(box_plot['boxes'], colors_list):
                patch.set_facecolor(color)
                patch.set_alpha(0.7)
            
            ax.set_title(f'Residual Distribution Comparison ({split.upper()} Set)', 
                        fontweight='bold', pad=20)
            ax.set_ylabel('Residual Value', fontweight='bold')
            ax.grid(True, alpha=0.3, axis='y')
            
            # Rotate x-axis labels for better readability
            ax.tick_params(axis='x', rotation=45)
        else:
            ax.text(0.5, 0.5, f'No data available for {split.upper()} set', 
                   transform=ax.transAxes, ha='center', va='center', fontsize=14)
            ax.set_title(f'{split.upper()} Set - No Data', fontweight='bold')
    
    plt.tight_layout()
    plt.show()

# Add this function to the existing code and call it after the individual analyses
def compare_models_analysis_publication(datasets, phenotypes, split='train'):
    """
    Publication-ready comparison analysis of pre-trained and non-pre-trained models
    Each plot is generated separately for easy integration into papers
    """
    
    for pheno in phenotypes:
        pred_col = f"{pheno}_pred"
        label_col = f"{pheno}_label"
        
        # Check if both models have data for this phenotype
        residuals_data = {}
        for model_name in ['pretrained', 'no_pretrain']:
            df = datasets[model_name][split]
            if not df.empty and pred_col in df.columns and label_col in df.columns:
                residuals = df[pred_col] - df[label_col]
                if not residuals.empty and residuals.nunique() >= 2:
                    residuals_data[model_name] = residuals
        
        if len(residuals_data) < 2:
            print(f"Skipping phenotype {pheno} for {split} set: Insufficient data")
            continue
            
        print(f"\n=== Phenotype {pheno.upper()} Comparison Analysis ({split.upper()} set) ===")
        
        # 1. Q-Q Plot Comparison
        fig1, ax1 = plt.subplots(figsize=(7, 5))
        
        # Get theoretical quantiles only once
        sample_size = min(len(residuals_data['pretrained']), len(residuals_data['no_pretrain']))
        theoretical_quantiles = scipy_stats.norm.ppf(np.linspace(0.01, 0.99, sample_size))
        
        # Plot sample quantiles for both models
        for model_name, residuals in residuals_data.items():
            sample_quantiles = np.sort(residuals)
            # Interpolate to match theoretical quantiles length
            if len(sample_quantiles) != len(theoretical_quantiles):
                x_interp = np.linspace(0, 1, len(sample_quantiles))
                x_new = np.linspace(0, 1, len(theoretical_quantiles))
                sample_quantiles = np.interp(x_new, x_interp, sample_quantiles)
            
            ax1.scatter(theoretical_quantiles, sample_quantiles, 
                       color=colors[model_name], alpha=0.7, s=20,
                       label=model_names[model_name], edgecolors='white', linewidth=0.5)
        
        # Add single reference line for normal distribution
        min_val = min(theoretical_quantiles.min(), 
                     min([np.sort(res)[0] for res in residuals_data.values()]))
        max_val = max(theoretical_quantiles.max(), 
                     max([np.sort(res)[-1] for res in residuals_data.values()]))
        ax1.plot([min_val, max_val], [min_val, max_val], 'k--', 
                alpha=0.8, linewidth=2, label='Normal Distribution')
        
        ax1.set_title(f'Q-Q Plot for {pheno.upper()} ({split.upper()})', fontweight='bold', pad=20)
        ax1.set_xlabel('Theoretical Quantiles', fontweight='bold')
        ax1.set_ylabel('Sample Quantiles', fontweight='bold')
        ax1.legend(frameon=True, fancybox=True, shadow=True)
        ax1.grid(True, alpha=0.3)
        plt.tight_layout()
        plt.show()
        
        # 2. Histogram Comparison with KDE
        fig2, ax2 = plt.subplots(figsize=(7, 5))
        
        for model_name, residuals in residuals_data.items():
            ax2.hist(residuals, bins=25, alpha=0.6, color=colors[model_name], 
                    label=model_names[model_name], density=True, edgecolor='white', linewidth=0.7)
            
            # Add KDE overlay
            kde_x = np.linspace(residuals.min(), residuals.max(), 100)
            kde = scipy_stats.gaussian_kde(residuals)
            ax2.plot(kde_x, kde(kde_x), color=colors[model_name], linewidth=2.5, alpha=0.8)
        
        ax2.set_title(f'Residual Distribution for {pheno.upper()} ({split.upper()})', fontweight='bold', pad=20)
        ax2.set_xlabel('Residual Value', fontweight='bold')
        ax2.set_ylabel('Density', fontweight='bold')
        ax2.legend(frameon=True, fancybox=True, shadow=True)
        ax2.grid(True, alpha=0.3)
        plt.tight_layout()
        plt.show()
        
        # 3. Box Plot Comparison (skip for individual analysis since we'll create combined version)
        if split == 'train':  # Only show individual box plots for training set
            fig3, ax3 = plt.subplots(figsize=(7, 5))
            
            box_data = [residuals_data['pretrained'], residuals_data['no_pretrain']]
            box_labels = ['Pre-trained', 'Non-pre-trained']
            
            box_plot = ax3.boxplot(box_data, labels=box_labels, patch_artist=True,
                                  boxprops=dict(linewidth=1.5),
                                  medianprops=dict(linewidth=2, color='black'),
                                  whiskerprops=dict(linewidth=1.5),
                                  capprops=dict(linewidth=1.5))
            
            # Set box colors
            box_plot['boxes'][0].set_facecolor(colors['pretrained'])
            box_plot['boxes'][0].set_alpha(0.7)
            box_plot['boxes'][1].set_facecolor(colors['no_pretrain'])
            box_plot['boxes'][1].set_alpha(0.7)
            
            ax3.set_title(f'Box Plot Comparison for {pheno.upper()} ({split.upper()})', fontweight='bold', pad=20)
            ax3.set_ylabel('Residual Value', fontweight='bold')
            ax3.grid(True, alpha=0.3, axis='y')
            plt.tight_layout()
            plt.show()
        
        # 4. Kernel Density Estimation Comparison
        fig4, ax4 = plt.subplots(figsize=(7, 5))
        
        for model_name, residuals in residuals_data.items():
            sns.kdeplot(data=residuals, ax=ax4, color=colors[model_name], 
                       label=model_names[model_name], linewidth=2.5, alpha=0.8)
        
        ax4.set_title(f'Kernel Density Estimation for {pheno.upper()} ({split.upper()})', fontweight='bold', pad=20)
        ax4.set_xlabel('Residual Value', fontweight='bold')
        ax4.set_ylabel('Density', fontweight='bold')
        ax4.legend(frameon=True, fancybox=True, shadow=True)
        ax4.grid(True, alpha=0.3)
        plt.tight_layout()
        plt.show()
        
        # Statistical Analysis Results
        print(f"\n--- Statistical Analysis Results for {split.upper()} Set ---")
        
        # Basic descriptive statistics
        print("\nDescriptive Statistics:")
        print("-" * 60)
        print(f"{'Metric':<20} {'Pre-trained':<15} {'Non-pre-trained':<15}")
        print("-" * 60)
        
        metrics = [
            ('Sample Size', lambda x: len(x)),
            ('Mean', lambda x: x.mean()),
            ('Std Dev', lambda x: x.std()),
            ('Variance', lambda x: x.var()),
            ('Skewness', lambda x: scipy_stats.skew(x)),
            ('Kurtosis', lambda x: scipy_stats.kurtosis(x)),
            ('Minimum', lambda x: x.min()),
            ('Maximum', lambda x: x.max()),
            ('Median', lambda x: x.median()),
            ('MAE', lambda x: x.abs().mean())
        ]
        
        for metric_name, metric_func in metrics:
            pre_val = metric_func(residuals_data['pretrained'])
            no_pre_val = metric_func(residuals_data['no_pretrain'])
            if metric_name == 'Sample Size':
                print(f"{metric_name:<20} {pre_val:<15.0f} {no_pre_val:<15.0f}")
            else:
                print(f"{metric_name:<20} {pre_val:<15.6f} {no_pre_val:<15.6f}")
        
        # Normality Tests
        print("\n\nNormality Tests:")
        print("-" * 60)
        
        for model_name, residuals in residuals_data.items():
            print(f"\n{model_names[model_name]} Model:")
            
            # Shapiro-Wilk Test
            if len(residuals) >= 3:
                shapiro_stat, shapiro_p = scipy_stats.shapiro(residuals)
                print(f"  Shapiro-Wilk Test:")
                print(f"    Statistic: {shapiro_stat:.6f}")
                print(f"    p-value: {shapiro_p:.6f}")
                print(f"    Result: {'Normal distribution' if shapiro_p > 0.05 else 'Non-normal distribution'}")
            
            # Kolmogorov-Smirnov Test
            residuals_normalized = (residuals - residuals.mean()) / residuals.std()
            ks_stat, ks_p = scipy_stats.kstest(residuals_normalized, 'norm')
            print(f"  Kolmogorov-Smirnov Test:")
            print(f"    Statistic: {ks_stat:.6f}")
            print(f"    p-value: {ks_p:.6f}")
            print(f"    Result: {'Normal distribution' if ks_p > 0.05 else 'Non-normal distribution'}")
        
        # Two-sample comparison tests
        pretrained_residuals = residuals_data['pretrained']
        no_pretrain_residuals = residuals_data['no_pretrain']
        
        print("\n\nTwo-Sample Comparison Tests:")
        print("-" * 60)
        
        # Mann-Whitney U Test (non-parametric)
        mw_stat, mw_p = scipy_stats.mannwhitneyu(pretrained_residuals, no_pretrain_residuals, 
                                                 alternative='two-sided')
        print(f"Mann-Whitney U Test:")
        print(f"  Statistic: {mw_stat:.6f}")
        print(f"  p-value: {mw_p:.6f}")
        print(f"  Result: {'No significant difference' if mw_p > 0.05 else 'Significant difference'} in distributions")
        
        # Levene's Test for Equal Variances
        levene_stat, levene_p = scipy_stats.levene(pretrained_residuals, no_pretrain_residuals)
        print(f"\nLevene's Test for Equal Variances:")
        print(f"  Statistic: {levene_stat:.6f}")
        print(f"  p-value: {levene_p:.6f}")
        print(f"  Result: {'Equal variances' if levene_p > 0.05 else 'Unequal variances'}")
        
        # t-test (parametric)
        if levene_p > 0.05:
            t_stat, t_p = scipy_stats.ttest_ind(pretrained_residuals, no_pretrain_residuals, 
                                               equal_var=True)
            test_name = "Student's t-test"
        else:
            t_stat, t_p = scipy_stats.ttest_ind(pretrained_residuals, no_pretrain_residuals, 
                                               equal_var=False)
            test_name = "Welch's t-test"
        
        print(f"\n{test_name}:")
        print(f"  Statistic: {t_stat:.6f}")
        print(f"  p-value: {t_p:.6f}")
        print(f"  Result: {'No significant difference' if t_p > 0.05 else 'Significant difference'} in means")
        
        # Effect Size (Cohen's d)
        pooled_std = np.sqrt(((len(pretrained_residuals) - 1) * pretrained_residuals.var() + 
                             (len(no_pretrain_residuals) - 1) * no_pretrain_residuals.var()) / 
                            (len(pretrained_residuals) + len(no_pretrain_residuals) - 2))
        cohens_d = (pretrained_residuals.mean() - no_pretrain_residuals.mean()) / pooled_std
        
        print(f"\nEffect Size (Cohen's d): {cohens_d:.6f}")
        
        if abs(cohens_d) < 0.2:
            effect_size = "Small"
        elif abs(cohens_d) < 0.8:
            effect_size = "Medium"
        else:
            effect_size = "Large"
        
        print(f"Effect Size Interpretation: {effect_size}")
        print("=" * 60)

# Execute comparison analysis for all splits
splits = ['train', 'val', 'test']

for split in splits:
    print(f"\n{'='*80}")
    print(f"ANALYZING {split.upper()} DATASET")
    print(f"{'='*80}")
    
    # Check if both models have data for this split
    has_data = False
    for model_name in ['pretrained', 'no_pretrain']:
        if not datasets[model_name][split].empty:
            has_data = True
            break
    
    if not has_data:
        print(f"No data available for {split} set")
        continue
    
    compare_models_analysis_publication(datasets, phenotypes, split=split)

# Create combined box plots for validation and test sets
print(f"\n{'='*80}")
print("COMBINED BOX PLOTS FOR VALIDATION AND TEST SETS")
print(f"{'='*80}")
plot_combined_boxplots(datasets, phenotypes, splits=['val', 'test'])
```

```python
def plot_combined_boxplots(datasets, phenotypes, splits=['val', 'test']):
    """
    Create separate box plots for validation and test sets across all phenotypes
    Each plot is generated separately for easy integration into papers
    """
    # Collect all residuals data
    all_residuals_data = {}
    
    for split in splits:
        all_residuals_data[split] = {}
        for pheno in phenotypes:
            pred_col = f"{pheno}_pred"
            label_col = f"{pheno}_label"
            
            residuals_data = {}
            for model_name in ['pretrained', 'no_pretrain']:
                df = datasets[model_name][split]
                if not df.empty and pred_col in df.columns and label_col in df.columns:
                    residuals = df[pred_col] - df[label_col]
                    if not residuals.empty and residuals.nunique() >= 2:
                        residuals_data[model_name] = residuals
            
            if len(residuals_data) == 2:  # Both models have data
                all_residuals_data[split][pheno] = residuals_data
    
    # Create separate plots for each split
    for split in splits:
        fig, ax = plt.subplots(figsize=(7, 6))  # Size for individual plot
        
        # Prepare data for box plot
        box_data = []
        box_labels = []
        colors_list = []
        
        for pheno in phenotypes:
            if pheno in all_residuals_data[split]:
                # Add pre-trained data
                box_data.append(all_residuals_data[split][pheno]['pretrained'])
                box_labels.append(f'{pheno.upper()}\nPre-trained')
                colors_list.append(colors['pretrained'])
                
                # Add non-pre-trained data
                box_data.append(all_residuals_data[split][pheno]['no_pretrain'])
                box_labels.append(f'{pheno.upper()}\nNon-pre-trained')
                colors_list.append(colors['no_pretrain'])
        
        if box_data:
            # Create box plot
            box_plot = ax.boxplot(box_data, labels=box_labels, patch_artist=True,
                                 boxprops=dict(linewidth=1.5),
                                 medianprops=dict(linewidth=2, color='black'),
                                 whiskerprops=dict(linewidth=1.5),
                                 capprops=dict(linewidth=1.5))
            
            # Set colors
            for patch, color in zip(box_plot['boxes'], colors_list):
                patch.set_facecolor(color)
                patch.set_alpha(0.7)
            
            ax.set_title(f'Residual Distribution Comparison ({split.upper()} Set)', 
                        fontweight='bold', pad=20)
            ax.set_ylabel('Residual Value', fontweight='bold')
            ax.grid(True, alpha=0.3, axis='y')
            
            # Rotate x-axis labels for better readability
            ax.tick_params(axis='x', rotation=45)
        else:
            ax.text(0.5, 0.5, f'No data available for {split.upper()} set', 
                   transform=ax.transAxes, ha='center', va='center', fontsize=14)
            ax.set_title(f'{split.upper()} Set - No Data', fontweight='bold')
        
        plt.tight_layout()
        plt.show()

def compare_models_analysis_publication(datasets, phenotypes, split='train'):
    """
    Publication-ready comparison analysis of pre-trained and non-pre-trained models
    Each plot is generated separately for easy integration into papers
    """
    
    for pheno in phenotypes:
        pred_col = f"{pheno}_pred"
        label_col = f"{pheno}_label"
        
        # Check if both models have data for this phenotype
        residuals_data = {}
        for model_name in ['pretrained', 'no_pretrain']:
            df = datasets[model_name][split]
            if not df.empty and pred_col in df.columns and label_col in df.columns:
                residuals = df[pred_col] - df[label_col]
                if not residuals.empty and residuals.nunique() >= 2:
                    residuals_data[model_name] = residuals
        
        if len(residuals_data) < 2:
            print(f"Skipping phenotype {pheno} for {split} set: Insufficient data")
            continue
            
        print(f"\n=== Phenotype {pheno.upper()} Comparison Analysis ({split.upper()} set) ===")
        
        # 1. Q-Q Plot Comparison
        fig1, ax1 = plt.subplots(figsize=(7, 5))
        
        # Get theoretical quantiles only once
        sample_size = min(len(residuals_data['pretrained']), len(residuals_data['no_pretrain']))
        theoretical_quantiles = scipy_stats.norm.ppf(np.linspace(0.01, 0.99, sample_size))
        
        # Plot sample quantiles for both models
        for model_name, residuals in residuals_data.items():
            sample_quantiles = np.sort(residuals)
            # Interpolate to match theoretical quantiles length
            if len(sample_quantiles) != len(theoretical_quantiles):
                x_interp = np.linspace(0, 1, len(sample_quantiles))
                x_new = np.linspace(0, 1, len(theoretical_quantiles))
                sample_quantiles = np.interp(x_new, x_interp, sample_quantiles)
            
            ax1.scatter(theoretical_quantiles, sample_quantiles, 
                       color=colors[model_name], alpha=0.7, s=20,
                       label=model_names[model_name], edgecolors='white', linewidth=0.5)
        
        # Add single reference line for normal distribution
        min_val = min(theoretical_quantiles.min(), 
                     min([np.sort(res)[0] for res in residuals_data.values()]))
        max_val = max(theoretical_quantiles.max(), 
                     max([np.sort(res)[-1] for res in residuals_data.values()]))
        ax1.plot([min_val, max_val], [min_val, max_val], 'k--', 
                alpha=0.8, linewidth=2, label='Normal Distribution')
        
        ax1.set_title(f'Q-Q Plot for {pheno.upper()} ({split.upper()})', fontweight='bold', pad=20)
        ax1.set_xlabel('Theoretical Quantiles', fontweight='bold')
        ax1.set_ylabel('Sample Quantiles', fontweight='bold')
        ax1.legend(frameon=True, fancybox=True, shadow=True)
        ax1.grid(True, alpha=0.3)
        plt.tight_layout()
        plt.show()
        
        # 2. Histogram Comparison with KDE
        fig2, ax2 = plt.subplots(figsize=(7, 5))
        
        for model_name, residuals in residuals_data.items():
            ax2.hist(residuals, bins=25, alpha=0.6, color=colors[model_name], 
                    label=model_names[model_name], density=True, edgecolor='white', linewidth=0.7)
            
            # Add KDE overlay
            kde_x = np.linspace(residuals.min(), residuals.max(), 100)
            kde = scipy_stats.gaussian_kde(residuals)
            ax2.plot(kde_x, kde(kde_x), color=colors[model_name], linewidth=2.5, alpha=0.8)
        
        ax2.set_title(f'Residual Distribution for {pheno.upper()} ({split.upper()})', fontweight='bold', pad=20)
        ax2.set_xlabel('Residual Value', fontweight='bold')
        ax2.set_ylabel('Density', fontweight='bold')
        ax2.legend(frameon=True, fancybox=True, shadow=True)
        ax2.grid(True, alpha=0.3)
        plt.tight_layout()
        plt.show()
        
        # 3. Box Plot Comparison (skip for individual analysis since we'll create combined version)
        if split == 'train':  # Only show individual box plots for training set
            fig3, ax3 = plt.subplots(figsize=(7, 5))
            
            box_data = [residuals_data['pretrained'], residuals_data['no_pretrain']]
            box_labels = ['Pre-trained', 'Non-pre-trained']
            
            box_plot = ax3.boxplot(box_data, labels=box_labels, patch_artist=True,
                                  boxprops=dict(linewidth=1.5),
                                  medianprops=dict(linewidth=2, color='black'),
                                  whiskerprops=dict(linewidth=1.5),
                                  capprops=dict(linewidth=1.5))
            
            # Set box colors
            box_plot['boxes'][0].set_facecolor(colors['pretrained'])
            box_plot['boxes'][0].set_alpha(0.7)
            box_plot['boxes'][1].set_facecolor(colors['no_pretrain'])
            box_plot['boxes'][1].set_alpha(0.7)
            
            ax3.set_title(f'Box Plot Comparison for {pheno.upper()} ({split.upper()})', fontweight='bold', pad=20)
            ax3.set_ylabel('Residual Value', fontweight='bold')
            ax3.grid(True, alpha=0.3, axis='y')
            plt.tight_layout()
            plt.show()
        
        # 4. Kernel Density Estimation Comparison
        fig4, ax4 = plt.subplots(figsize=(7, 5))
        
        for model_name, residuals in residuals_data.items():
            sns.kdeplot(data=residuals, ax=ax4, color=colors[model_name], 
                       label=model_names[model_name], linewidth=2.5, alpha=0.8)
        
        ax4.set_title(f'Kernel Density Estimation for {pheno.upper()} ({split.upper()})', fontweight='bold', pad=20)
        ax4.set_xlabel('Residual Value', fontweight='bold')
        ax4.set_ylabel('Density', fontweight='bold')
        ax4.legend(frameon=True, fancybox=True, shadow=True)
        ax4.grid(True, alpha=0.3)
        plt.tight_layout()
        plt.show()
        
        # Statistical Analysis Results
        print(f"\n--- Statistical Analysis Results for {split.upper()} Set ---")
        
        # Basic descriptive statistics
        print("\nDescriptive Statistics:")
        print("-" * 60)
        print(f"{'Metric':<20} {'Pre-trained':<15} {'Non-pre-trained':<15}")
        print("-" * 60)
        
        metrics = [
            ('Sample Size', lambda x: len(x)),
            ('Mean', lambda x: x.mean()),
            ('Std Dev', lambda x: x.std()),
            ('Variance', lambda x: x.var()),
            ('Skewness', lambda x: scipy_stats.skew(x)),
            ('Kurtosis', lambda x: scipy_stats.kurtosis(x)),
            ('Minimum', lambda x: x.min()),
            ('Maximum', lambda x: x.max()),
            ('Median', lambda x: x.median()),
            ('MAE', lambda x: x.abs().mean())
        ]
        
        for metric_name, metric_func in metrics:
            pre_val = metric_func(residuals_data['pretrained'])
            no_pre_val = metric_func(residuals_data['no_pretrain'])
            if metric_name == 'Sample Size':
                print(f"{metric_name:<20} {pre_val:<15.0f} {no_pre_val:<15.0f}")
            else:
                print(f"{metric_name:<20} {pre_val:<15.6f} {no_pre_val:<15.6f}")
        
        # Normality Tests
        print("\n\nNormality Tests:")
        print("-" * 60)
        
        for model_name, residuals in residuals_data.items():
            print(f"\n{model_names[model_name]} Model:")
            
            # Shapiro-Wilk Test
            if len(residuals) >= 3:
                shapiro_stat, shapiro_p = scipy_stats.shapiro(residuals)
                print(f"  Shapiro-Wilk Test:")
                print(f"    Statistic: {shapiro_stat:.6f}")
                print(f"    p-value: {shapiro_p:.6f}")
                print(f"    Result: {'Normal distribution' if shapiro_p > 0.05 else 'Non-normal distribution'}")
            
            # Kolmogorov-Smirnov Test
            residuals_normalized = (residuals - residuals.mean()) / residuals.std()
            ks_stat, ks_p = scipy_stats.kstest(residuals_normalized, 'norm')
            print(f"  Kolmogorov-Smirnov Test:")
            print(f"    Statistic: {ks_stat:.6f}")
            print(f"    p-value: {ks_p:.6f}")
            print(f"    Result: {'Normal distribution' if ks_p > 0.05 else 'Non-normal distribution'}")
        
        # Two-sample comparison tests
        pretrained_residuals = residuals_data['pretrained']
        no_pretrain_residuals = residuals_data['no_pretrain']
        
        print("\n\nTwo-Sample Comparison Tests:")
        print("-" * 60)
        
        # Mann-Whitney U Test (non-parametric)
        mw_stat, mw_p = scipy_stats.mannwhitneyu(pretrained_residuals, no_pretrain_residuals, 
                                                 alternative='two-sided')
        print(f"Mann-Whitney U Test:")
        print(f"  Statistic: {mw_stat:.6f}")
        print(f"  p-value: {mw_p:.6f}")
        print(f"  Result: {'No significant difference' if mw_p > 0.05 else 'Significant difference'} in distributions")
        
        # Levene's Test for Equal Variances
        levene_stat, levene_p = scipy_stats.levene(pretrained_residuals, no_pretrain_residuals)
        print(f"\nLevene's Test for Equal Variances:")
        print(f"  Statistic: {levene_stat:.6f}")
        print(f"  p-value: {levene_p:.6f}")
        print(f"  Result: {'Equal variances' if levene_p > 0.05 else 'Unequal variances'}")
        
        # t-test (parametric)
        if levene_p > 0.05:
            t_stat, t_p = scipy_stats.ttest_ind(pretrained_residuals, no_pretrain_residuals, 
                                               equal_var=True)
            test_name = "Student's t-test"
        else:
            t_stat, t_p = scipy_stats.ttest_ind(pretrained_residuals, no_pretrain_residuals, 
                                               equal_var=False)
            test_name = "Welch's t-test"
        
        print(f"\n{test_name}:")
        print(f"  Statistic: {t_stat:.6f}")
        print(f"  p-value: {t_p:.6f}")
        print(f"  Result: {'No significant difference' if t_p > 0.05 else 'Significant difference'} in means")
        
        # Effect Size (Cohen's d)
        pooled_std = np.sqrt(((len(pretrained_residuals) - 1) * pretrained_residuals.var() + 
                             (len(no_pretrain_residuals) - 1) * no_pretrain_residuals.var()) / 
                            (len(pretrained_residuals) + len(no_pretrain_residuals) - 2))
        cohens_d = (pretrained_residuals.mean() - no_pretrain_residuals.mean()) / pooled_std
        
        print(f"\nEffect Size (Cohen's d): {cohens_d:.6f}")
        
        if abs(cohens_d) < 0.2:
            effect_size = "Small"
        elif abs(cohens_d) < 0.8:
            effect_size = "Medium"
        else:
            effect_size = "Large"
        
        print(f"Effect Size Interpretation: {effect_size}")
        print("=" * 60)

# Execute comparison analysis for all splits
splits = ['train', 'val', 'test']

for split in splits:
    print(f"\n{'='*80}")
    print(f"ANALYZING {split.upper()} DATASET")
    print(f"{'='*80}")
    
    # Check if both models have data for this split
    has_data = False
    for model_name in ['pretrained', 'no_pretrain']:
        if not datasets[model_name][split].empty:
            has_data = True
            break
    
    if not has_data:
        print(f"No data available for {split} set")
        continue
    
    compare_models_analysis_publication(datasets, phenotypes, split=split)

# Create separate box plots for validation and test sets
print(f"\n{'='*80}")
print("SEPARATE BOX PLOTS FOR VALIDATION AND TEST SETS")
print(f"{'='*80}")
plot_combined_boxplots(datasets, phenotypes, splits=['val', 'test'])
```
