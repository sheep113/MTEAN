```python
import pickle
import os
from pathlib import Path

# 假设脚本在 Whisper_of_DNA_pl 目录下运行
# 或者根据实际情况调整基础路径
base_path = Path(__file__).parent.parent if '__file__' in locals() else Path.cwd()
base_path = base_path.parent
file_path = base_path / "data" / "example-data" / "maize1404" / "SNP_pca" / "1404cubic.pkl"

# 检查文件是否存在
if not file_path.exists():
    print(f"错误：文件未找到 - {file_path}")
else:
    try:
        # 以二进制读取模式打开文件
        with open(file_path, 'rb') as f:
            # 加载 pickle 文件中的数据
            data = pickle.load(f)

        # 现在 'data' 变量包含了 pkl 文件中的内容
        print(f"成功加载数据从: {file_path}")

        # 你可以检查数据的类型和内容
        print(f"数据类型: {type(data)}")
        # 如果数据是字典或类似结构，可以打印键
        if isinstance(data, dict):
            print(f"数据键: {list(data.keys())}")
        # 如果数据是 numpy 数组，可以打印形状
        elif hasattr(data, 'shape'):
            print(f"数据形状: {data.shape}")
        else:
            # 打印部分数据以供预览
            print(f"数据预览 (前 100 个字符): {str(data)[:100]}")

    except FileNotFoundError:
        print(f"错误：文件未找到 - {file_path}")
    except pickle.UnpicklingError:
        print(f"错误：无法反序列化文件 - {file_path}. 文件可能已损坏或不是有效的 pickle 文件。")
    except Exception as e:
        print(f"加载文件时发生未知错误: {e}")
```
